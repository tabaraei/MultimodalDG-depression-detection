```python
from src.dataset_loader import AndroidsCorpusDataset
from IPython.display import Audio
import torch

device = torch.device('cuda:0')
FOLD_IDX = 0


def inspect_data(dataset, PARTICIPANT_IDX=0, SEGMENT_IDX=0):
    X_audio, X_text, y = dataset[PARTICIPANT_IDX]
    print(X_text[SEGMENT_IDX])
    return Audio(X_audio[SEGMENT_IDX], rate=dataset.SAMPLE_RATE)
```
